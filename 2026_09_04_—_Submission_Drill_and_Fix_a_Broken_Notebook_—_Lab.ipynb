### How to read an error before you fix it

Python's traceback is read **bottom up**. The last line names the exception and the problem; the line above it points at your code.

```
Traceback (most recent call last):
  File "<stdin>", line 2, in <module>
    print(greeting)
NameError: name 'greeting' is not defined
```

Four exceptions cover most of what you will hit this semester:

| Exception | Usually means |
|---|---|
| `NameError` | The variable was never created, or the cell that created it never ran |
| `TypeError` | You mixed types — text where a number belongs is the classic |
| `KeyError` | That column or dictionary key does not exist. Check spelling and case |
| `FileNotFoundError` | The path is wrong for *this* environment |

### Bug 6 — the answer changes when you re-run it

Run the next cell three times and watch the total move. Then fix it using what you built Wednesday: leave the source data alone and derive a new column from it. Your fix must give the same total on every run.

````python
sales = pd.DataFrame({
    'game': ['NC State', 'Louisville'],
    'units': [1450, 1610],
})

# Derive a new column so re-running cannot change the answer
sales['adjusted_units'] = sales['units'] * 1.05
print('total units:', round(sales['adjusted_units'].sum(), 2))
````

````python
# Proof — this must print the same number three times
for _ in range(3):
    # Rebuild from the raw values and print your total
    sales_proof = pd.DataFrame({
        'game': ['NC State', 'Louisville'],
        'units': [1450, 1610],
    })
    sales_proof['adjusted_units'] = sales_proof['units'] * 1.05
    print('total units:', round(sales_proof['adjusted_units'].sum(), 2))
````

**What was wrong:** The original 'units' column was modified in place (`sales['units'] = sales['units'] * 1.05`), causing the total to change every time the cell was re-run. The fix involves creating a new column (`adjusted_units`) for the calculation, leaving the source 'units' column unchanged.

### Bug 7 — a path from someone else's laptop

This notebook was written on a machine you do not have. Fix it with the pattern from Wednesday's studio: look for the file, and fall back to the inline sample so the notebook still runs for whoever opens it next.

````python
import os
from io import StringIO
import pandas as pd # Import pandas as it's used in this cell

SAMPLE = '''game,units\nNC State,1450\nLouisville,1610\n'''

# Fallback logic: Try to read from a file, if not found, use the sample data.
file_path = 'gameday_orders.csv' # Assuming the file might be in the current directory if provided
if os.path.exists(file_path):
    orders = pd.read_csv(file_path)
else:
    orders = pd.read_csv(StringIO(SAMPLE))

orders
````

**What was wrong:** The code tried to read a file from a hardcoded absolute path (`/Users/someone/Desktop/gameday_orders.csv`) that did not exist in the current environment. The fix involves implementing a fallback mechanism: attempting to read from a local file first, and if that fails, reading from an inline `SAMPLE` string to ensure the notebook runs anywhere.
